1. Using PandasDataFrame
 Product of first two column


```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1,10, size=(5,3)), columns=['A','B','C'])
df['D'] = df['A'] * df['B']
print(df)
```

2.Row-wise and Column-wise Sum

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1,10, size=(4,3)), columns=['A','B','C'])
print(df)
print("\n")
df.loc['Col_Sum'] = df.sum()
df['Row_Sum'] = df.sum(axis=1)
print(df)
```

3.Filling NaN with mean

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1,20, size=(5,3)),
                  columns=['A','B','C'])
df.iloc[1,0] = np.nan
df.iloc[3,1] = np.nan

#before
print(df)
print("\n")
#after
df_filled = df.fillna(df.mean())
print(df_filled)
```

4.Droping the rows with NaN

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1,20, size=(6,4)),
                  columns=['A','B','C','D'])
df.iloc[2,1] = np.nan
df.iloc[4,2] = np.nan
print("Before dropna")
print(df)
print("\n")
print("After drop")
df_dropped = df.dropna()
print(df_dropped)
```

5.Data Aggregation(Sum and Mean)

```python

import pandas as pd
import numpy as np
df = pd.DataFrame({
    'Category': np.random.choice(['A','B','C'], 10),
    'Value': np.random.randint(10,100,10)
})
print(df)
print("\n")

result = df.groupby('Category')['Value'].agg(['sum','mean'])
print(result)
```

6.Group by category and total sales

```python
df = pd.DataFrame({
    'Product': ['P1','P2','P3','P4','P5','P6'],
    'Category': np.random.choice(['Electronics','Food','Cloth'], 6),
    'Sales': np.random.randint(100,1000,6)
})
print(df)
print("\n")
total_sales = df.groupby('Category')['Sales'].sum()
print(total_sales)
```

7.Merging dataframe

```python

import pandas as pd
df1 = pd.DataFrame({
    'ID':[1,2,3],
    'Name':['A','B','C']
    })
df2 = pd.DataFrame({
    'ID':[1,2,3],
    'Marks':[90,85,88]
    })
merged = pd.merge(df1, df2, on='ID')
print(merged)
```

8. MultiIndex Dataframe

```python
import pandas as pd

index = pd.MultiIndex.from_tuples([
    ('A', 1), ('A', 2),
    ('B', 1), ('B', 2)
])

df = pd.DataFrame({'Value': [10, 20, 30, 40]},
                  index=index)

print(df)

print("Indexing:")
print(df.loc['A'])

print("Slicing:")
print(df.loc[('A', 1):('B', 1)])
```

9.MultIndex(category and subcategory sum)

```python
import pandas as pd
import numpy as np
index = pd.MultiIndex.from_product([['A','B','C'], ['X','Y']], names=['Category','SubCategory'])
df = pd.DataFrame(np.random.randint(10,50, size=(6,2)), index=index, columns=['Val1','Val2'])

sum_by_cat = df.groupby(level='Category').sum()
print(sum_by_cat)
print("\n")
sum_by_both = df.groupby(level=['Category','SubCategory']).sum()
print(sum_by_both)
```

10.Apply function to Double values

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1,10, size=(5,3)),
                                    columns=['A','B','C'])
print(df)
df_doubled = df.apply(lambda x: x*2) # or df*2
print(df_doubled)
```

11.lambda function to create sum columns

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1,10, size=(6,3)),
                  columns=['A','B','C'])
df['Total'] = df.apply(lambda row: row['A']+row['B']+row['C'], axis=1)
print(df)
```
